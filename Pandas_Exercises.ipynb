### Exercise A8: Fill the gaps  &nbsp; `⭐⭐ Medium`

Make a copy called `clean` and fill its missing values like this:
- **City** → the text `'Unknown'`
- **Maths** → the **mean** of the Maths column
- **Science** → the **median** of the Science column

Then confirm with `clean.isnull().sum()`. Which column still has a gap?

```python
# Write your code here
# Create a copy of the original DataFrame
clean = students.copy()

# Fill missing City values
clean['City'] = clean['City'].fillna('Unknown')

# Fill missing Maths values with the mean
clean['Maths'] = clean['Maths'].fillna(clean['Maths'].mean())

# Fill missing Science values with the median
clean['Science'] = clean['Science'].fillna(clean['Science'].median())

# Check remaining missing values
clean.isnull().sum()
```

<details><summary>💡 Hint</summary>

Start with `clean = students.copy()`. Then, e.g., `clean['Maths'] = clean['Maths'].fillna(clean['Maths'].mean())`.

</details>

<details><summary>✅ Check your answer</summary>

Only **English** still has **1** missing value (we did not fill it yet). Maths was filled with about **79.71**, Science with **74.0**.

</details>

### Exercise B8: Friendly column names (named aggregation)  &nbsp; `⭐⭐ Medium`

Build a **City summary** table with these three columns:

### Exercise B9: Group by two columns  &nbsp; `⭐⭐⭐ Challenge`

Which City + Category pair earns the most?

```python
# Write your code here
city_category_revenue = (
    orders.groupby(['City', 'Category'])['Revenue']
    .sum()
    .reset_index()
    .sort_values(by='Revenue', ascending=False)
)

print(city_category_revenue)
```

# Part C: Concat, Merge and Join 🔗

**Story:** Data is rarely in one place. Monthly sales sit in separate files, and customers, orders and products live in different tables.
You will practise stacking tables (**concat**) and matching tables using a key (**merge** / **join**).

Run the cell below to create all the tables.

```python
# --- Monthly sales files (same columns) ---
jan = pd.DataFrame({'OrderID': [1, 2, 3],  'Customer': ['Asha', 'Ravi', 'Meena'],  'Amount': [2500, 1800, 3200]})
feb = pd.DataFrame({'OrderID': [4, 5],     'Customer': ['Kiran', 'Divya'],         'Amount': [2100, 2750]})
mar = pd.DataFrame({'OrderID': [6, 7, 8],  'Customer': ['Asha', 'Meena', 'Ravi'],  'Amount': [1500, 2900, 3300]})
apr = pd.DataFrame({'OrderID': [9, 10],    'Customer': ['Kiran', 'Asha'],          'Amount': [4000, 2200],
                    'Discount': [200, 0]})   # April has an EXTRA column!

# --- Three linked tables ---
customers = pd.DataFrame({
    'CustomerID': [1, 2, 3, 4, 5],
    'Name':       ['Asha', 'Ravi', 'Meena', 'Kiran', 'Divya'],
    'City':       ['Hyderabad', 'Pune', 'Delhi', 'Chennai', 'Hyderabad'],
})
orders2 = pd.DataFrame({
    'OrderID':    [501, 502, 503, 504, 505, 506],
    'CustomerID': [1, 2, 1, 3, 6, 2],          # customer 6 is NOT in the customers table!
    'ProductID':  [101, 102, 101, 103, 102, 101],
    'Qty':        [1, 2, 1, 3, 1, 2],
})
products = pd.DataFrame({
    'ProductID':   [101, 102, 103, 104],
    'ProductName': ['Laptop', 'Mouse', 'Keyboard', 'Monitor'],
    'Price':       [55000, 500, 1500, 12000],
})

# --- Tables for the multi-column merge ---
sales   = pd.DataFrame({'Month': ['Jan', 'Jan', 'Feb', 'Feb'], 'Store': ['A', 'B', 'A', 'B'], 'Sales':  [100, 150, 120, 130]})
targets = pd.DataFrame({'Month': ['Jan', 'Jan', 'Feb', 'Mar'], 'Store': ['A', 'B', 'A', 'A'], 'Target': [90, 160, 110, 140]})

display(orders2)
display(customers)
display(products)
```

### Quick reference
| I want to... | Code |
|---|---|
| Count missing values | `df.isnull().sum()` |
| Drop rows with gaps | `df.dropna()` |
| Fill gaps | `df.fillna(value)` · `df['col'].fillna(df['col'].mean())` · `df.ffill()` |
| Total per group | `df.groupby('A')['B'].sum()` |
| Several stats per group | `df.groupby('A')['B'].agg(['sum', 'mean'])` |
| Stack tables | `pd.concat([a, b], ignore_index=True)` |
| Match tables by a key | `pd.merge(a, b, on='key', how='left')` |
| Match tables by index | `a.join(b)` |
